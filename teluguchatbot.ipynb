```python
import gradio as gr
from googletrans import Translator

translator = Translator()

def telugu_agent(user_input, target_word):
    # 1. Translate the target to Telugu
    correct_telugu = translator.translate(target_word, dest='te').text

    # 2. Check the user's answer
    if user_input.strip() == correct_telugu:
        return f"🎉 Correct! '{target_word}' is indeed '{correct_telugu}'."
    else:
        return f"❌ Not quite. '{target_word}' in Telugu is '{correct_telugu}'. Try again!"

# Create a Simple UI
demo = gr.Interface(
    fn=telugu_agent,
    inputs=["text", gr.Textbox(placeholder="Enter an English word to learn")],
    outputs="text",
    title="Telugu Learning Agent 🇮🇳",
    description="Type an English word and its Telugu translation to test yourself!"
)

demo.launch()
```

```python
!pip install googletrans==4.0.0-rc1 gradio

import gradio as gr
from googletrans import Translator

# 1. Setup our Data (Our Dictionary)
telugu_emojis = {
    "🍎": "Apple",
    "🏠": "House",
    "☀️": "Sun",
    "🍛": "Food",
    "🚗": "Car"
}

translator = Translator()

# 2. The Logic Engine (The Agent)
def telugu_learning_hub(emoji_choice, user_guess):
    # Get the English word from our dictionary
    english_word = telugu_emojis.get(emoji_choice, "Unknown")

    # Translate it to Telugu automatically
    # This ensures the bot always stays 'Technical' and 'Logic-driven'
    translation = translator.translate(english_word, dest='te')
    correct_telugu = translation.text
    phonetic = translation.pronunciation if translation.pronunciation else ""

    # Check the user's answer
    # We check if their guess is in the English word or the Telugu script
    if user_guess.lower().strip() in [english_word.lower(), correct_telugu.strip()]:
        return (f"🎉 Correct! \n\n"
                f"Symbol: {emoji_choice} \n"
                f"English: {english_word} \n"
                f"Telugu: {correct_telugu} ({phonetic})")
    else:
        return (f"❌ Not quite! \n\n"
                f"For {emoji_choice}, the word is '{correct_telugu}'.\n"
                f"Keep practicing!")

# 3. Create a Professional UI
with gr.Blocks() as demo:
    gr.Markdown("# 🇮🇳 Telugu Emoji Agent")
    gr.Markdown("Select an emoji and try to guess the word in English or Telugu!")

    with gr.Row():
        emoji_input = gr.Dropdown(choices=list(telugu_emojis.keys()), label="Pick an Emoji")
        guess_input = gr.Textbox(placeholder="Your guess here...", label="Guess the Word")

    output_text = gr.Textbox(label="Agent Feedback")

    submit_btn = gr.Button("Check Answer")
    submit_btn.click(fn=telugu_learning_hub, inputs=[emoji_input, guess_input], outputs=output_text)

# 4. Launch it!
demo.launch()
```

```python
# Our Telugu Learning Dictionary
telugu_emojis = {
    "🍎": "ఆపిల్ (Apple)",
    "🏠": "ఇల్లు (Illu)",
    "☀️": "సూర్యుడు (Sooryudu)",
    "🍛": "అన్నం (Annam)"
}

def check_emoji_answer(emoji_key, user_guess):
    correct_word = telugu_emojis[emoji_key]
    if user_guess.lower() in correct_word.lower():
        return f"✅ Correct! {emoji_key} is {correct_word}"
    else:
        return f"❌ Try again! {emoji_key} is {correct_word}"

print(check_emoji_answer("🍎", "apple"))
```
